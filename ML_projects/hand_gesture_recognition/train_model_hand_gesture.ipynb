```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv('data.csv')
df.columns
```

```python
# df.drop(columns=['rel_hull_area' , 'rel_cont_area'] , inplace = True)
```

```python

from sklearn.model_selection import train_test_split

X = df[['rel_cont_area' , 'rel_hull_area' , 'num_defects','sum_angles']]
y = df[['thumb' , 'index' ,'middle' , 'ring' , 'pinky']]

X_train , X_test  , y_train , y_test  = train_test_split(X ,y , test_size=0.2 , stratify = y)
len(X_test)
```

```python
from sklearn.preprocessing import StandardScaler , MinMaxScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_train.to_numpy())
from sklearn.linear_model import LogisticRegression 
from sklearn.multiclass import OneVsRestClassifier , OneVsOneClassifier
lr = OneVsRestClassifier(LogisticRegression(penalty='l1' , solver = 'saga', max_iter=1000))
lr.fit(X_scaled , y_train)
```

```python

from sklearn.metrics import precision_score , recall_score , accuracy_score
y_predict = lr.predict(scaler.transform(X_test)) 
prec = precision_score(y_test ,y_predict , average ='macro' )
rec =  recall_score(y_test ,y_predict , average ='macro' )
acc= accuracy_score(y_predict , y_test)
print(prec , rec , acc)
 
```

```python

from joblib import dump 
dump(lr ,'hand_gesture_model.pkl' )
 
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score ,precision_score , recall_score
from sklearn.preprocessing import StandardScaler , MinMaxScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_train.to_numpy())
knn = KNeighborsClassifier(n_neighbors = 4 , weights = 'distance' , n_jobs=-1)
knn.fit(X_scaled , y_train)
```

```python

y_pred = knn.predict(scaler.transform(X_test))
prec = precision_score(y_test , y_pred , average = 'macro')
rec = recall_score(y_test , y_pred , average = 'macro')
acc = accuracy_score(y_test , y_pred )
print(prec , rec , acc)
 
```
